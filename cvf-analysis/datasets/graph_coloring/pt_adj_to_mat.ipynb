```python
import numpy as np
```

```python
f = open("graph_random_regular_graph_n4_d3_pt_adj_list.txt", "r")
```

```python
N = 256
```

```python
matrix = np.zeros((N, N))
```

```python
line = f.readline()
while line:
    i = 0
    line_split = line.split(",")
    while i + 1 < len(line_split):
        matrix[int(line_split[i]), int(line_split[i+1])] = 1 
        matrix[int(line_split[i+1]), int(line_split[i])] = 1
        i += 1
    line = f.readline()
```

```python
matrix.shape
```

```python
import pickle
```

```python
pickle.dump(matrix, open("matrix.dump", "wb"))
```
