```python
import networkx as nx

graph = "star_graph_n3"

# create empty graph
G = nx.Graph()

# open file and read line by line
with open(f"{graph}.txt", "r") as f:
    for line in f:
        numbers = list(map(int, line.strip().split()))
        node = numbers[0]         # first number is the node
        neighbors = numbers[1:]   # rest are neighbors
        
        for neighbor in neighbors:
            G.add_edge(node, neighbor)

# check
print("Nodes:", G.nodes())
print("Edges:", G.edges())
```

```python
def print_dict_sorted_by_values(d: dict):
    for key, value in sorted(d.items(), key=lambda item: item[1], reverse=True):
        print(f"{key}: {value}")
```

```python
def print_dict_sorted(d: dict):
    for key in sorted(d.keys()):
        print(f"{d[key]}")
```

```python
degree_centrality = nx.degree_centrality(G)
print_dict_sorted(degree_centrality), print(), print_dict_sorted_by_values(degree_centrality)
```
