# Generate powerlaw cluster graph

```python
import networkx as nx

# Parameters
n = 30000 # Number of nodes
m = 4 # Number of edges to attach from a new node to existing nodes
p = 0.3 # Probability of adding a triangle after adding a random edge

# Generate the power-law cluster graph
G = nx.powerlaw_cluster_graph(n, m, p)

file_name = f"graph_powerlaw_cluster_graph_n{n}_m{m}_p{int(p*100)}"
# Write graph to a text file
with open(f"{file_name}.txt", "w") as f:
    for node in G.nodes():
        neighbors = sorted(G.neighbors(node))
        line = f"{node} " + " ".join(str(neigh) for neigh in neighbors)
        f.write(line + "\n")

print(f"Graph saved to '{file_name}.txt'")
```

# Generate star graph

```python
import networkx as nx

# Parameters
n = 30000 # Number of nodes

# Generate the power-law cluster graph
G = nx.star_graph(n-1)

file_name = f"graph_star_graph_n{n}"
# Write graph to a text file
with open(f"{file_name}.txt", "w") as f:
    for node in G.nodes():
        neighbors = sorted(G.neighbors(node))
        line = f"{node} " + " ".join(str(neigh) for neigh in neighbors)
        f.write(line + "\n")

print(f"Graph saved to '{file_name}.txt'")
```

# Generate random regular graph

```python
import networkx as nx

# Parameters
n = 10  # number of nodes
d = 4   # degree of each node

# Generate a random d-regular graph
G = nx.random_regular_graph(d, n)

# Write graph to a text file in adjacency list format
file_name = f"graph_random_regular_graph_n{n}_d{d}"
with open(f"{file_name}.txt", "w") as f:
    for node in G.nodes():
        neighbors = sorted(G.neighbors(node))
        line = f"{node} " + " ".join(str(neigh) for neigh in neighbors)
        f.write(line + "\n")

print(f"Random regular graph saved to '{file_name}.txt'")
```

# Generate complete graph

```python
import networkx as nx

n = 50
G = nx.complete_graph(n)

file_name = f"complete_graph_n{n}"
with open(f"{file_name}.txt", "w") as f:
    for node in G.nodes():
        neighbors = sorted(G.neighbors(node))
        line = f"{node} " + " ".join(str(neigh) for neigh in neighbors)
        f.write(line + "\n")

print(f"Complete graph saved to '{file_name}.txt'")
```

# Generate complete bipartite graph

```python
import networkx as nx


m = 4
n = 4

G = nx.complete_bipartite_graph(m, n)

file_name = f"complete_bipartite_graph_m{m}_n{n}"
with open(f"{file_name}.txt", "w") as f:
    for node in G.nodes():
        neighbors = sorted(G.neighbors(node))
        line = f"{node} " + " ".join(str(neigh) for neigh in neighbors)
        f.write(line + "\n")

print(f"Complete bipartite graph saved to '{file_name}.txt'")
```
