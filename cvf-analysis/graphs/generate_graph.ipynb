```python
import networkx as nx

# Parameters
n = 3
m = 3
p = 0.3

# Generate the power-law cluster graph
G = nx.powerlaw_cluster_graph(n, m, p)

file_name = f"graph_powerlaw_cluster_graph_n{n}"
# Write graph to a text file
with open(f"{file_name}.txt", "w") as f:
    for node in G.nodes():
        neighbors = sorted(G.neighbors(node))
        line = f"{node} " + " ".join(str(neigh) for neigh in neighbors)
        f.write(line + "\n")

print(f"Graph saved to '{file_name}.txt'")
```

```python
import networkx as nx

# Parameters
n = 6  # number of nodes
d = 2  # degree of each node

# Generate a random d-regular graph
G = nx.random_regular_graph(d, n)

# Write graph to a text file in adjacency list format
file_name = f"graph_random_regular_graph_n{n}_d{d}"
with open(f"{file_name}.txt", "w") as f:
    for node in G.nodes():
        neighbors = sorted(G.neighbors(node))
        line = f"{node} " + " ".join(str(neigh) for neigh in neighbors)
        f.write(line + "\n")

print(f"Random regular graph saved to '{file_name}.txt'")
```
